```python
transactions = [
    (100,"local"),
    (200,"local"),
    (1000,"local"),
    (2000,"foreign"),
    (200,"local"),
]
MAX_NORMAL_AMOUNT = 1000
AlLOWED_LOCATIONS = ["local"]
def detect_anomalies(transactions):
  anomalies = []
  for idx,(amount,location) in enumerate(transactions):
    if amount > MAX_NORMAL_AMOUNT or location not in AlLOWED_LOCATIONS:
      anomalies.append({
          "transaction_id": idx,
          "amount":amount,
          "location":location,
          "reason":("High Amount" if amount > MAX_NORMAL_AMOUNT else "") + ("Unusual Location" if location not in AlLOWED_LOCATIONS else "")
                })
  return anomalies
anomalies = detect_anomalies(transactions)
if anomalies:
  print("Suspicious transactions detected:")
  for anomaly in anomalies:
    print(f"ID{anomaly['transaction_id']}: ${anomaly['amount']} at {anomaly['location']} -> {anomaly['reason']}")
else:
  print("No anomalies detected.")
```

```python
import pandas as pd

# Assuming you have a CSV file named 'transactions.csv' with 'amount' and 'location' columns
# For demonstration, let's create a dummy CSV file first
data = [
    [150, 'local'],
    [1200, 'foreign'],
    [300, 'local'],
    [800, 'remote'],
    [2500, 'local'],
]
df_transactions = pd.DataFrame(data, columns=['amount', 'location'])
df_transactions.to_csv('transactions.csv', index=False)

# Now, load the data from the CSV file
loaded_transactions_df = pd.read_csv('transactions.csv')
print("Data loaded from CSV:")
print(loaded_transactions_df.head())

# You would then convert this DataFrame back to the list of tuples format
# that your detect_anomalies function expects if you wanted to use it directly:
loaded_transactions_list = list(loaded_transactions_df.itertuples(index=False, name=None))

print("\nAnomalies in loaded data (if any):")
anomalies_from_csv = detect_anomalies(loaded_transactions_list)

if anomalies_from_csv:
    for anomaly in anomalies_from_csv:
        print(f"ID{anomaly['transaction_id']}: ${anomaly['amount']} at {anomaly['location']} -> {anomaly['reason']}")
else:
    print("No anomalies detected in the data loaded from CSV.")
```

```python
import random

# Generate a larger dataset for demonstration
def generate_large_transactions(num_transactions):
    large_transactions = []
    for i in range(num_transactions):
        amount = random.randint(50, 5000) # Vary amounts
        location = random.choice(["local", "foreign", "remote"])
        large_transactions.append((amount, location))
    return large_transactions

# Generate 100 transactions
new_transactions = generate_large_transactions(100)

print(f"Generated {len(new_transactions)} new transactions.")

# Detect anomalies with the new dataset
new_anomalies = detect_anomalies(new_transactions)

if new_anomalies:
  print("\nSuspicious transactions detected in the new dataset:")
  for anomaly in new_anomalies:
    print(f"ID{anomaly['transaction_id']}: ${anomaly['amount']} at {anomaly['location']} -> {anomaly['reason']}")
else:
  print("No anomalies detected in the new dataset.")
```
