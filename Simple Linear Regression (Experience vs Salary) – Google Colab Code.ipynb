```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
df = pd.read_csv("Experience-Salary.csv")

# Show first 5 rows
df.head()
```

```python
df.rename(columns={
    "exp(in months)": "experience",
    "salary(in thousands)": "salary"
}, inplace=True)

df.info()
```

```python
df.describe()
```

```python
X = df[["experience"]]   # Independent feature
y = df["salary"]         # Dependent feature
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.figure(figsize=(12, 6))

sns.scatterplot(x=X_train["experience"], y=y_train, label="Training Data")
plt.plot(X_test["experience"], y_pred, color="red", label="Regression Line")

plt.xlabel("Experience (Months)")
plt.ylabel("Salary (Thousands)")
plt.title("Experience vs Salary - Linear Regression")
plt.legend()
plt.show()
```

```python
print("Coefficient (Slope):", model.coef_)
print("Intercept:", model.intercept_)
```
