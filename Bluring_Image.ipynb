Bluring the image

```python
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

im = Image.open("/content/google.webp")

im1 = im.filter(ImageFilter.BoxBlur(10))

plt.imshow(im1)
plt.title("Blured image")
plt.axis("on")
```

Providing Tint Color

```python
from PIL import ImageEnhance
im = Image.open("/content/google.webp").convert("RGB")
blue_tint = Image.new("RGB", im.size, (255, 0, 255))
tinted = Image.blend(im, blue_tint, alpha= 0.5)
plt.imshow(tinted)
plt.title("Blured image")
plt.axis("on")
```
