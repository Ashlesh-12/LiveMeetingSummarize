```python

with open('sample.txt','w') as f:
  f.write("This text is going to convert into audio")
```

```python
!pip install gtts

from gtts import gTTS
from IPython.display import display,Audio
text = open("sample.txt").read()
tts = gTTS(text=text,lang = 'en')
tts.save("sample_audio.mp3")

display(Audio('sample_audio.mp3',autoplay=False))
```

```python
import subprocess

subprocess.run(
    ["ffmpeg", "-y", "-i", "sample_audio.mp3", "-ar", "16000", "-ac", "1", "audio.wav"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

print("Converted to audio.wav (16k mono)")
```
