```python
import subprocess

subprocess.run([
    "ffmpeg", "-y",
    "-i", "audio.ogg",
    "-ar", "16000",
    "-ac", "1",
    "audio.wav"
], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
```

```python
import whisper
import torch

model = whisper.load_model("small")

# Load audio
audio = whisper.load_audio("audio.wav")
audio = whisper.pad_or_trim(audio)

# Create log-Mel spectrogram
mel = whisper.log_mel_spectrogram(audio).to(model.device)

# Detect language
_, probs = model.detect_language(mel)

# Sort probabilities
sorted_langs = sorted(probs.items(), key=lambda x: x[1], reverse=True)

print("Top language predictions:")
for lang, prob in sorted_langs[:5]:
    print(f"{lang} → {prob:.3f}")
```

```python
best_lang = sorted_langs[0][0]
print("\nDetected Language Code:", best_lang)
```
