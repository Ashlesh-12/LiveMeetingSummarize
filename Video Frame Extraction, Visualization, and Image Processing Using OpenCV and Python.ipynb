```python
import cv2
import os

video_path = "/content/Devasena.mp4"
output_folder = "frames_all"

os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_filename = os.path.join(output_folder, f"frame_{frame_count:03d}.jpg")
    cv2.imwrite(frame_filename, frame)
    frame_count += 1

cap.release()
cv2.destroyAllWindows()

print(f"✅ Saved {frame_count} frames")
```

```python
import cv2
import os

video_path = "/content/Devasena.mp4"
output_folder = "frames_n"

os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

frame_count = 0
max_frames = 10   # 👈 change as needed

while frame_count < max_frames:
    ret, frame = cap.read()
    if not ret:
        break

    frame_filename = os.path.join(output_folder, f"frame_{frame_count:03d}.jpg")
    cv2.imwrite(frame_filename, frame)
    frame_count += 1

cap.release()
cv2.destroyAllWindows()

print(f"✅ Saved {frame_count} frames")
```

```python
from IPython.display import display, HTML
import os, base64

folder_path = "frames_n"
N = 6

html = ""

for file in sorted(os.listdir(folder_path))[:N]:
    with open(os.path.join(folder_path, file), "rb") as f:
        encoded = base64.b64encode(f.read()).decode()

    html += f"""
    <img src="data:image/jpeg;base64,{encoded}"
         width="200"
         style="margin:6px;border:1px solid #ccc;">
    """

display(HTML(html))
```

```python
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

image_path = "/content/Screenshot 2025-06-13 141822.png"

im = Image.open(image_path).convert("RGB")

# Apply blur
im_blur = im.filter(ImageFilter.BoxBlur(40))

plt.imshow(im_blur)
plt.title("Blurred Image")
plt.axis("off")
plt.show()
```
