```python
from nltk.stem import WordNetLemmatizer
import nltk

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

print("stones:", lemmatizer.lemmatize("stones"))
print("speaking:", lemmatizer.lemmatize("speaking"))
print("bedroom:", lemmatizer.lemmatize("bedroom"))
print("cats:", lemmatizer.lemmatize("cats"))
```

```python
from nltk.stem import SnowballStemmer

snow = SnowballStemmer("english")

words = ["running", "studies", "playing", "caring", "happily", "cats"]

for w in words:
    print(w, snow.stem(w))
```

```python
import nltk
from nltk.stem import PorterStemmer

nltk.download('punkt')
nltk.download('punkt_tab') # Added to download the missing resource

ps = PorterStemmer()
file_path = "stemming.txt"

with open(file_path, 'r') as file:
    text = file.read()

words = nltk.word_tokenize(text)
stemmed_words = [ps.stem(word) for word in words]

print("Original Text:\n", text)
print("\nStemmed Output:\n")
print(" ".join(stemmed_words))
```
