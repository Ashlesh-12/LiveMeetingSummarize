```python
import pandas as pd
```

```python
df=pd.read_csv('data/all_kindle_review.csv')
```

```python
df.head()
```

```python
df=df[['reviewText','rating']]
```

```python
df.head()
```

```python
df.info()
```

```python
def rating(rating):
    if rating<=2:
        return -1
    elif rating==3:
        return 0
    else:
        return 1
```

```python
df["rating"]=df["rating"].apply(lambda x: rating(x))
```

```python
df.head()
```

```python
df.to_csv('data/kindle_sentiment.csv',index=False)
```
