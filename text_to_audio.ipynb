```python
import os
import base64
from datetime import datetime
from gtts import gTTS
from io import BytesIO
```

```python
# Create directory for storing audio files
os.makedirs('audio_outputs', exist_ok=True)
```

```python
def get_audio_as_base64(audio_bytes):
    """Convert audio bytes to base64 for download"""
    return base64.b64encode(audio_bytes).decode()
```

```python
def generate_speech(text, language='en', slow=False):
    """
    Generate speech from text using gTTS

    Parameters:
    - text: Text to convert to speech
    - language: Language code (default: 'en' for English)
    - slow: Boolean to slow down speech (default: False)

    Returns:
    - audio_bytes: Audio in MP3 format
    """
    try:
        tts = gTTS(text=text, lang=language, slow=slow, tld='co.in')
        audio_buffer = BytesIO()
        tts.write_to_fp(audio_buffer)
        audio_buffer.seek(0)
        return audio_buffer.getvalue()
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
# Title and Description
print("🎙️ EchoVocal - Text to Audio Converter")
print("=" * 50)
print("Transform your text into crystal-clear speech instantly.")
print("Powered by gTTS -  No API Keys Required")
print("=" * 50)

# User Input
text_input = input("\n📝 Enter your text (max 500 characters): ").strip()

if text_input:
    if len(text_input) > 500:
        print("❌ Text exceeds 500 characters. Please enter shorter text.")
    else:
        # Language Selection
        print("\n🌍 Available Languages:")
        print("1. English (en)")
        print("2. Hindi (hi)")
        print("3. Spanish (es)")
        print("4. French (fr)")
        print("5. German (de)")
        print("6. Japanese (ja)")

        lang_choice = input("Select language (1-6, default: 1): ").strip()

        lang_map = {
            '1': 'en',
            '2': 'hi',
            '3': 'es',
            '4': 'fr',
            '5': 'de',
            '6': 'ja'
        }

        selected_lang = lang_map.get(lang_choice, 'en')

        # Speed Selection
        slow_choice = input("Slow down speech? (y/n, default: n): ").strip().lower()
        slow = slow_choice == 'y'

        # Generate Speech
        print("\n⏳ Generating speech...")
        audio_data = generate_speech(text_input, language=selected_lang, slow=slow)

        if isinstance(audio_data, str) and "Error" in audio_data:
            print(f"❌ {audio_data}")
        else:
            # Save Audio File
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            filename = f"audio_outputs/speech_{timestamp}.mp3"

            with open(filename, 'wb') as f:
                f.write(audio_data)

            print(f"✅ Audio generated successfully!")
            print(f"📁 Saved as: {filename}")
            print(f"📊 File size: {len(audio_data) / 1024:.2f} KB")

            # Display Audio Player
            print("\n🔊 Audio Player (Click to play):")
            audio_base64 = get_audio_as_base64(audio_data)
            print(f"[Audio file ready for download: {filename}]")
else:
    print("⚠️ Please enter some text to convert to speech.")
```
