```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
print("stones:", lemmatizer.lemmatize("stones"))
print("speaking:", lemmatizer.lemmatize("speaking"))
print("bedroom:", lemmatizer.lemmatize("bedroom"))
print("cats:", lemmatizer.lemmatize("cats"))
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
words = ["walking", "cooking", "playing", "sharing", "happily", "cats"]
for w in words:
    print(w, "→", ps.stem(w))
```
