```python
from groq import Groq
from google.colab import userdata
import numpy as np
import scipy.io.wavfile as wav

# Retrieve the API key from Colab's Secrets Manager
GROQ_API_KEY = userdata.get('GROQ_API_KEY')
client = Groq(api_key=GROQ_API_KEY)

# Using the correct file name based on available files
file_path = "/content/inputaudio.wav"

try:
    # Transcribe the audio file
    with open(file_path, "rb") as audio_file:
        response = client.audio.transcriptions.create(
            file=audio_file,
            model="whisper-large-v3",
            response_format="verbose_json"
        )

    print("Language:", response.language)
    print("Text:", response.text)
    print("-" * 50)
except FileNotFoundError:
    print(f"Error: Audio file '{file_path}' not found. Please ensure it is uploaded to your Colab environment.")
except Exception as e:
    print(f"An error occurred during transcription: {e}")
```
