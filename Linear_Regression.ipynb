```python
import numpy
import matplotlib.pyplot as mtp
import pandas as pd
```

```python
data_set = pd.read_csv("/content/Salary_Data.csv")
print(data_set.to_string())
```

```python
x = data_set.iloc[:, : -1]
y = data_set.iloc[:, 1].values
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 1/3, random_state = 0)
print(x_test)
print(y_test)
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train, y_train)
```

```python
x_pred = regressor.predict(x_test)
print(x_pred)
y_pred = regressor.predict(x_test)
print(y_pred)
```

```python
mtp.scatter(x_train, y_train, color = "red")
mtp.plot(x_train, regressor.predict(x_train), color = "green")
mtp.title("Salary vs Experience (Training set)")
mtp.xlabel("Years of Experience")
mtp.ylabel("Salary")
mtp.show()
```

```python
mtp.scatter(x_test, y_test, color = "red")
mtp.plot(x_train, regressor.predict(x_train), color = "green")
mtp.title("Salary vs Experience (Testing set)")
mtp.xlabel("Years of Experience")
mtp.ylabel("Salary")
mtp.show()
```
